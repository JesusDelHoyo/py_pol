```python
from py_pol.stokes import Stokes
from scipy.spatial import ConvexHull
```

```python
S = Stokes().Fibonacci_spiral(N=5)[0:-1]
hull = ConvexHull(S.M[1:,:].T)
volume = hull.volume
area = hull.area
```

```python
# Espiral para N=5 y quito el ultimo punto
print(f"Volume: {volume:.3f}, Area: {area:.3f}, ratio: {volume/area:.3f}")
S.draw_poincare(colormap='hot',kind='scatterline')
```

```python
S2 = Stokes().circular_light(length=4)
Saux = Stokes().Fibonacci_spiral(N=3)
S2[0:3] = Saux
hull = ConvexHull(S2.M[1:,:].T)
volume = hull.volume
area = hull.area
```

```python
# Añadido el polo
print(f"Volume: {volume:.3f}, Area: {area:.3f}, ratio: {volume/area:.3f}")
S2.draw_poincare(colormap='hot',kind='scatterline')
```
